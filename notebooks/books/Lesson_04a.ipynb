## 1. HK5's Drill Problem 1-13

In this problem, we use short circuits (the s element) for the first time — four of them, one for each current the problem asks about. Find i1, i2, i3 and i4, on the current through line of each short's block.

![The circuit of HK5's Drill Problem 1-13](https://learn.symbulator.com/assets/circuit/hk5d0113.jpg)

```python
c1 = '''
r1,1,0,25
jd,0,2,0.2*v1
r2,2,3,10
ji,4,3,2.5
r3,4,5,100
s1,1,2
s2,2,4
s3,0,3
s4,3,5
'''
draw(c1)
```

```python
r1 = dc(c1)
r1
```

## 2. AS2's Practice Problem 2.15

Find the equivalent resistance as seen by the 100 V source, and the current i. The equivalent resistance comes free here: a source sits between the two nodes, so re is already an answer.

![The circuit of AS2's Practice Problem 2.15](https://learn.symbulator.com/assets/circuit/as2pp0215.jpg)

```python
c2 = '''
e,a,0,100
r13,a,1,13
r24,1,2,24
r10,1,3,10
r20,2,3,20
r30,2,0,30
r50,3,0,50
'''
draw(c2)
```

## 3. B11's Example 8.29 (step by step)

In this problem, we use the Find equivalent analysis (the calculator's er script) for the first time, on a passive circuit. Find the equivalent resistance between nodes 0 and c — the lesson walks this one through step by step.

![The circuit of B11's Example 8.29 (step by step)](https://learn.symbulator.com/assets/circuit/b11e0829.jpg)

```python
c3 = '''
r4,0,a,4
r2,0,b,2
r6,a,b,6
rb,a,c,3
ra,b,c,3
'''
draw(c3)
```

```python
r3 = er(c3, '0', 'c', domain='dc')
r3
```

## 4. RM3's Practice Problem 9-4

In this problem, we find a Thévenin / Norton equivalent (the calculator's th script) for the first time. Find the equivalent seen from nodes 2 and 0: vth, ino, req and pmax all arrive together.

![The circuit of RM3's Practice Problem 9-4](https://learn.symbulator.com/assets/circuit/rm3pp0904.jpg)

## 5. B11's Example 9.6

Find the Thévenin equivalent of the shaded network, then the current through RL for values of 2, 10 and 100 ohm. The load question comes ticked, so irl appears under the four answers; the Evaluate line reads it at the 2 ohm load, 1.5 A; change the condition to load = 10 and load = 100 for the other two: 0.5 A and 0.059 A. Rounding is approx to 2 digits, as the chapter says, so the answers read as the book prints them.

![The circuit of B11's Example 9.6](https://learn.symbulator.com/assets/circuit/b11e0906.jpg)

```python
c5 = '''
e,1,0,9
r1,1,2,3
r2,2,0,6
'''
draw(c5)
```

```python
r5 = th(c5, '2', '0', domain='dc')
r5
```

```python
evaluate(r5, 'irl', conditions=['load = 2'])
```

## 6. RM3's Example 9-8

Find the Norton equivalent of the circuit left of a-b; the load question comes ticked, ready for the Load circuit equivalent button, which writes the next entry. Version 9 reports the Norton current in the direction it flows, so the sign differs from the calculator's 0.36.

![The circuit of RM3's Example 9-8](https://learn.symbulator.com/assets/circuit/rm3e0908.jpg)

```python
c6 = '''
e,1,0,24
r1,1,2,120
r2,2,0,280
j,2,0,560'm
'''
draw(c6)
```

```python
r6 = th(c6, '2', '0', domain='dc')
r6
```

## 7. RM3's Example 9-8 (with the load)

The equivalent from the previous entry, as the Load circuit equivalent button writes it, with the load changed to 168 ohm and the 180 mA source added. Find the current through RL: with SI prefixes on, irl reads -60 mA, that is, 60 mA flowing from 0 to n.

![The circuit of RM3's Example 9-8 (with the load)](https://learn.symbulator.com/assets/circuit/rm3e0908.jpg)

```python
c7 = '''
jN,0,n,iNo
rE,n,0,rEq
rL,n,0,168
j,0,n,180'm
'''
draw(c7)
```

```python
r7 = dc(c7, conditions=['iNo=-9/25', 'rEq=84'])
r7.rounded(3)
```
