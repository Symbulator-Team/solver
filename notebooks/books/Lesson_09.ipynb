## 1. AS7's Examples 12.2 & 12.6 (balanced wye-wye)

Calculate the line currents of the three-wire Y-Y system, and the complex power absorbed at the source and at the load; the values are RMS. The centre of both wyes can be node 0 only because the system is balanced. Complex power at the source is -2086.2 - 834.5j VA, and at the load sra0+srb0+src0 = 1390.8 + 1112.6j VA.

![The circuit of AS7's Examples 12.2 & 12.6 (balanced wye-wye)](https://learn.symbulator.com/assets/circuit/as7e1202.png)

```python
c1 = '''
ea0,ag,0,(110∠0°)
eb0,bg,0,(110∠-120°)
ec0,cg,0,(110∠120°)
rat,ag,ad,5-2j
rbt,bg,bd,5-2j
rct,cg,cd,5-2j
ra0,ad,0,10+8j
rb0,bd,0,10+8j
rc0,cd,0,10+8j
'''
draw(c1)
```

```python
r1 = ac(c1, omega='omega', use_rms=True)
r1.rounded(4)
```

```python
evaluate(r1, 'sea0+seb0+sec0')
```
