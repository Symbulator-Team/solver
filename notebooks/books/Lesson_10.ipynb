## 1. AS7's Example 13.1 (mutual inductance in jOhm)

In this problem, we describe a mutual inductance (the m element) for the first time. Calculate the phasor currents I1 and I2 — aa(ir2) and aa(ir4). The coils are given in jOhm, so they are r elements with imaginary values and the mutual matches. The dotted end of each coupled element is named first.

![The circuit of AS7's Example 13.1 (mutual inductance in jOhm)](https://learn.symbulator.com/assets/circuit/as7e1301.png)

```python
c1 = '''
e,1,0,12
r1,1,2,-4j
r2,2,0,5j
m,r2,r3,3j
r3,3,0,6j
r4,3,0,12
'''
draw(c1)
```

```python
r1 = ac(c1, omega='omega')
r1.rounded(4)
```

## 8. The autotransformer as one tapped winding

## 9. A transformer whose primary is not grounded

The 8 ohm load reflects to the primary as 32 ohm, in series with 1 + 3, so the primary current is 5/18 A: it2 = 5/18 entering at node 2, it3 = -5/18 leaving at node 3, and it4 = -5/9 on the secondary, doubled by the 2:1 ratio and negative because it leaves the transformer into the load. irl = 5/9 A.

## 10. Both windings between live nodes, and why a side needs a ground

Both windings between live nodes: the 100 ohm load sits across the secondary and node 5 is tied to ground through 7 ohm. it2 = 5/202, it4 = -5/202, it3 = -5/101, it5 = 5/101, and ir5 = 0: the secondary's current circulates through the load alone. The resistor is what gives the secondary side a reference -- an ideal transformer conducts nothing from one winding to the other. Delete r5 and Symbulator still solves, taking node 5 as that side's own reference and saying so in a note; the currents and v3 - v5 come out the same.

![The circuit of Both windings between live nodes, and why a side needs a ground](https://learn.symbulator.com/assets/circuit/sym_t_live_pairs.png)

```python
c10 = '''
e,1,0,10
r0,1,2,1
t,[2,4],[3,5],[2,1]
r4,4,0,3
rl,3,5,100
r5,5,0,7
'''
draw(c10)
```

```python
r10 = dc(c10)
r10.rounded(4)
```

## 12. NR11's 60 V source and coupled 2 H and 8 H coils (TR, t > 0)
