## 1. B11's Example 5.7

The first simulation in the tutorial: a series circuit, described and solved step by step in Lesson 1. Find the resistance seen by the source, the source current, the voltage drop across each resistor, the power delivered by the source and the power consumed by each resistor. The Evaluate line checks the power balance: the resistors consume exactly what the source delivers, so the four powers come to zero.

![The circuit of B11's Example 5.7](https://learn.symbulator.com/assets/circuit/b11e0507.jpg)

```python
c1 = '''
e,1,0,36
r1,1,2,1'k
r2,2,3,3'k
r3,3,0,2'k
'''
draw(c1)
```

```python
r1 = dc(c1)
r1.rounded(3)
```
